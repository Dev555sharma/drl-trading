```python
# Cell 2 — create env and take a few random steps
env = TradingEnvironment(split="train")
state = env.reset()

print(f"State shape : {state.shape}")     # should be (102,)
print(f"Max steps   : {env.max_steps}")   # should be ~1565 for train split
print(f"Balance     : ${env.balance:,.2f}")
```

```python
# Cell 3 — simulate 5 random actions
import numpy as np

for step in range(5):
    action = np.random.choice([0, 1, 2])   # random: hold / buy / sell
    action_name = ["Hold", "Buy", "Sell"][action]

    next_state, reward, done = env.step(action)

    print(f"Step {step+1} | Action: {action_name:4s} | "
          f"Reward: {reward:+.4f} | "
          f"Net worth: ${env.net_worth:,.2f}")
```
